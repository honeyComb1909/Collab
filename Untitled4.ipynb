```python

# ============================================================
# PRACTICAL 6: IMAGE SEGMENTATION USING LIGHTWEIGHT U-NET
# Automatic dataset | 3 epochs | CPU-friendly | No Google Drive
# ============================================================

%pip -q install tensorflow-datasets importlib_resources

import os
import time
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import tensorflow_datasets as tfds

from tensorflow.keras import layers, Model
from google.colab import files
from PIL import Image

# ---------------- CONFIGURATION ----------------
IMG_SIZE = 96
BATCH_SIZE = 16
EPOCHS = 3
NUM_CLASSES = 3
AUTOTUNE = tf.data.AUTOTUNE
SEED = 42

tf.keras.utils.set_random_seed(SEED)

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
print("Epochs:", EPOCHS)

# ---------------- AUTOMATIC DATASET DOWNLOAD ----------------
print("\nDownloading Oxford-IIIT Pet dataset...")

(train_raw, val_raw, test_raw), info = tfds.load(
    "oxford_iiit_pet",
    split=["train[:90%]", "train[90%:]", "test"],
    with_info=True,
    data_dir="/content/tfds_pet_data",
    download=True
)

print("Dataset downloaded successfully.")

# ---------------- PREPROCESSING ----------------
def preprocess(example):
    image = tf.image.resize(
        example["image"],
        (IMG_SIZE, IMG_SIZE),
        method="bilinear"
    )
    image = tf.cast(image, tf.float32) / 255.0

    mask = tf.image.resize(
        example["segmentation_mask"],
        (IMG_SIZE, IMG_SIZE),
        method="nearest"
    )
    mask = tf.cast(tf.squeeze(mask, axis=-1), tf.int32)

    # Original mask labels 1, 2, 3 -> 0, 1, 2
    mask = tf.clip_by_value(mask - 1, 0, 2)
    return image, mask

train_ds = (
    train_raw.map(preprocess, num_parallel_calls=AUTOTUNE)
    .shuffle(500, seed=SEED)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

val_ds = (
    val_raw.map(preprocess, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

test_ds = (
    test_raw.map(preprocess, num_parallel_calls=AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(AUTOTUNE)
)

# ---------------- LIGHTWEIGHT U-NET ----------------
def conv_block(x, filters):
    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    x = layers.Conv2D(
        filters, 3, padding="same", activation="relu"
    )(x)
    return x

def build_unet():
    inputs = layers.Input((IMG_SIZE, IMG_SIZE, 3))

    # Encoder
    c1 = conv_block(inputs, 16)
    p1 = layers.MaxPooling2D()(c1)

    c2 = conv_block(p1, 32)
    p2 = layers.MaxPooling2D()(c2)

    # Bottleneck
    b = conv_block(p2, 64)

    # Decoder
    u2 = layers.Conv2DTranspose(
        32, 2, strides=2, padding="same"
    )(b)
    u2 = layers.Concatenate()([u2, c2])
    c3 = conv_block(u2, 32)

    u1 = layers.Conv2DTranspose(
        16, 2, strides=2, padding="same"
    )(c3)
    u1 = layers.Concatenate()([u1, c1])
    c4 = conv_block(u1, 16)

    outputs = layers.Conv2D(
        NUM_CLASSES, 1, activation="softmax"
    )(c4)

    return Model(inputs, outputs, name="Lightweight_U_Net")

model = build_unet()

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["sparse_categorical_accuracy"]
)

model.summary()

# ---------------- TRAIN MODEL: ONLY 3 EPOCHS ----------------
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=1,
        restore_best_weights=True,
        verbose=1
    )
]

print("\nTraining model...")
start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks
)

print(f"Training time: {(time.time()-start_time)/60:.2f} minutes")

# ---------------- ACCURACY AND LOSS GRAPHS ----------------
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(
    history.history["sparse_categorical_accuracy"],
    marker="o", label="Training"
)
axes[0].plot(
    history.history["val_sparse_categorical_accuracy"],
    marker="o", label="Validation"
)
axes[0].set_title("Pixel Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[0].grid(True)

axes[1].plot(history.history["loss"], marker="o", label="Training")
axes[1].plot(history.history["val_loss"], marker="o", label="Validation")
axes[1].set_title("Loss")
axes[1].set_xlabel("Epoch")
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()

# ---------------- TEST EVALUATION ----------------
print("\nEvaluating on the test dataset...")
test_loss, test_accuracy = model.evaluate(test_ds, verbose=1)

print("\n========== FINAL TEST RESULTS ==========")
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Pixel Accuracy: {test_accuracy * 100:.2f}%")
print("Target: 85-90% (not guaranteed)")
print("========================================")

# ---------------- MEAN INTERSECTION OVER UNION ----------------
miou = tf.keras.metrics.MeanIoU(num_classes=NUM_CLASSES)

for images, masks in test_ds:
    probabilities = model(images, training=False)
    predictions = tf.argmax(
        probabilities, axis=-1, output_type=tf.int32
    )
    miou.update_state(masks, predictions)

print(f"Mean IoU: {miou.result().numpy() * 100:.2f}%")

# ---------------- DISPLAY SAMPLE PREDICTIONS ----------------
images, masks = next(iter(test_ds))
predictions = model(images[:4], training=False)
predictions = tf.argmax(predictions, axis=-1).numpy()

fig, axes = plt.subplots(4, 3, figsize=(10, 12))

for i in range(4):
    axes[i, 0].imshow(images[i].numpy())
    axes[i, 0].set_title("Original")

    axes[i, 1].imshow(
        masks[i].numpy(), cmap="viridis", vmin=0, vmax=2
    )
    axes[i, 1].set_title("Ground Truth")

    axes[i, 2].imshow(
        predictions[i], cmap="viridis", vmin=0, vmax=2
    )
    axes[i, 2].set_title("U-Net Prediction")

    for j in range(3):
        axes[i, j].axis("off")

plt.tight_layout()
plt.show()

# ---------------- SAVE MODEL ----------------
MODEL_PATH = "/content/unet_pet_segmentation.keras"
model.save(MODEL_PATH)

print("\nModel saved:", MODEL_PATH)

# ---------------- UPLOAD YOUR OWN IMAGE ----------------
print("\nUpload an image from your computer.")
print("Supported formats: JPG, JPEG, PNG")

uploaded = files.upload()

for filename in uploaded.keys():
    try:
        original = Image.open(filename).convert("RGB")
        resized = original.resize((IMG_SIZE, IMG_SIZE))

        image_array = (
            np.asarray(resized, dtype=np.float32) / 255.0
        )
        image_array = np.expand_dims(image_array, axis=0)

        probabilities = model.predict(image_array, verbose=0)[0]
        predicted_mask = np.argmax(
            probabilities, axis=-1
        ).astype(np.uint8)

        fig, axes = plt.subplots(1, 2, figsize=(11, 5))

        axes[0].imshow(original)
        axes[0].set_title("Uploaded Image")
        axes[0].axis("off")

        axes[1].imshow(
            predicted_mask, cmap="viridis", vmin=0, vmax=2
        )
        axes[1].set_title("Predicted Segmentation")
        axes[1].axis("off")

        plt.tight_layout()
        plt.show()

        mask_path = "/content/predicted_segmentation.png"
        Image.fromarray(predicted_mask * 127).save(mask_path)

        print("Downloading segmentation mask...")
        files.download(mask_path)

    except Exception as error:
        print(f"Could not process {filename}: {error}")

print("\nPRACTICAL 6 COMPLETED SUCCESSFULLY")
```
